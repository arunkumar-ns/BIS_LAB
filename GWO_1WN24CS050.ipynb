```python

import random

traffic = {
    "North": 40,
    "South": 30,
    "East": 60,
    "West": 50
}

directions = list(traffic.keys())

MIN_TIME = 10
MAX_TIME = 60
TOTAL_TIME = 120

def objective_function(position):
    total_waiting_time = 0

    for i in range(4):
        green_time = position[i]
        vehicles = traffic[directions[i]]

        waiting_time = vehicles * (TOTAL_TIME - green_time)
        total_waiting_time += waiting_time

    return total_waiting_time

def initialize_population(num_wolves):
    population = []

    for _ in range(num_wolves):
        wolf = [
            random.uniform(MIN_TIME, MAX_TIME)
            for _ in range(4)
        ]
        population.append(wolf)

    return population

def fitness(position):
    return objective_function(position)

def update_position(wolf, alpha, beta, delta, a):
    new_position = []

    for i in range(4):
        A1 = 2 * a * random.random() - a
        C1 = 2 * random.random()
        D_alpha = abs(C1 * alpha[i] - wolf[i])
        X1 = alpha[i] - A1 * D_alpha

        A2 = 2 * a * random.random() - a
        C2 = 2 * random.random()
        D_beta = abs(C2 * beta[i] - wolf[i])
        X2 = beta[i] - A2 * D_beta

        A3 = 2 * a * random.random() - a
        C3 = 2 * random.random()
        D_delta = abs(C3 * delta[i] - wolf[i])
        X3 = delta[i] - A3 * D_delta

        value = (X1 + X2 + X3) / 3
        value = max(MIN_TIME, min(MAX_TIME, value))

        new_position.append(value)

    total = sum(new_position)

    new_position = [
        (value / total) * TOTAL_TIME
        for value in new_position
    ]

    return new_position

def main():
    random.seed(42)

    num_wolves = 20
    max_iterations = 100

    population = initialize_population(num_wolves)

    for iteration in range(max_iterations):

        population.sort(key=fitness)

        alpha = population[0]
        beta = population[1]
        delta = population[2]

        a = 2 - iteration * (2 / max_iterations)

        new_population = []

        for wolf in population:
            new_wolf = update_position(
                wolf, alpha, beta, delta, a
            )
            new_population.append(new_wolf)

        population = new_population

    population.sort(key=fitness)

    best_solution = population[0]
    best_fitness = fitness(best_solution)

    print("Optimal Traffic Signal Timings")

    for i in range(4):
        print(
            directions[i], ":",
            round(best_solution[i], 2), "seconds"
        )

    print("Total Waiting Time:", round(best_fitness, 2))

main()
```

```python

import numpy as np
import random

cities = {
    "Depot": (10, 10),
    "City A": (20, 30),
    "City B": (40, 20),
    "City C": (50, 50),
    "City D": (30, 60),
    "City E": (10, 50)
}

city_names = list(cities.keys())
coordinates = np.array(list(cities.values()))
n = len(cities)

def distance(city1, city2):
    return np.linalg.norm(coordinates[city1] - coordinates[city2])

def Objective_function(route):
    total = 0
    for i in range(n):
        total += distance(route[i], route[(i + 1) % n])
    return total

def Levy_Flight(solution):
    new_solution = solution.copy()
    i, j = random.sample(range(1, n), 2)
    new_solution[i], new_solution[j] = new_solution[j], new_solution[i]
    return new_solution

def Evaluate_Fitness(solution):
    return Objective_function(solution)

def Abandon_Nests(nests, fitness, Pa):
    count = max(1, int(Pa * len(nests)))
    worst_indices = np.argsort(fitness)[-count:]

    for i in worst_indices:
        new_nest = list(range(1, n))
        random.shuffle(new_nest)
        nests[i] = [0] + new_nest

    return nests

def best_solution(nests):
    fitness = [Evaluate_Fitness(nest) for nest in nests]
    index = np.argmin(fitness)
    return nests[index].copy(), fitness[index]

def CS(Objective_function, N, Pa, ITER):
    nests = []

    for i in range(N):
        nest = list(range(1, n))
        random.shuffle(nest)
        nests.append([0] + nest)

    fitness = [Evaluate_Fitness(nest) for nest in nests]
    best, best_fitness = best_solution(nests)

    for iteration in range(ITER):
        for i in range(N):
            new_solution = Levy_Flight(nests[i])
            new_fitness = Evaluate_Fitness(new_solution)

            j = random.randrange(N)

            if new_fitness < fitness[j]:
                nests[j] = new_solution
                fitness[j] = new_fitness

        nests = Abandon_Nests(nests, fitness, Pa)
        fitness = [Evaluate_Fitness(nest) for nest in nests]

        best, best_fitness = best_solution(nests)

    return best, best_fitness

N = 25
Pa = 0.25
ITER = 200

Best_solution, Best_fitness = CS(
    Objective_function, N, Pa, ITER
)

print("Optimized Delivery Route:")

route = Best_solution + [Best_solution[0]]

for city in Best_solution:
    print(city_names[city], end=" -> ")

print(city_names[Best_solution[0]])

print("\nShortest Distance:",
      round(Best_fitness, 2), "units")

print("======================================================================")
print("Arun Kumar N S")
print("1WN24CS050")
```

```python
# Question:
# Implement Ant Colony Optimization (ACO) to solve the Traveling
# Salesman Problem (TSP) for a tourist route optimization system.
# Use hard-coded locations, find a short route visiting every
# location exactly once, return to the starting point, and display
# the optimized route and total distance.

import math
import random

cities = {
    "Hotel": (10, 10),
    "Museum": (15, 20),
    "Park": (5, 25),
    "Beach": (25, 30),
    "Mall": (30, 10)
}

NUM_ANTS = 15
ITERATIONS = 50
ALPHA = 1.0
BETA = 2.0
RHO = 0.5
Q = 100.0

random.seed(42)

names = list(cities.keys())
n = len(names)

distance = [[0.0] * n for _ in range(n)]

for i in range(n):
    for j in range(n):
        x1, y1 = cities[names[i]]
        x2, y2 = cities[names[j]]
        distance[i][j] = math.hypot(x1 - x2, y1 - y2)


def initialize_pheromone():
    return [
        [1.0 if i != j else 0.0 for j in range(n)]
        for i in range(n)
    ]


pheromone = initialize_pheromone()


def select_next_city(current, unvisited):
    candidates = sorted(unvisited)
    weights = [
        pheromone[current][city] ** ALPHA *
        (1 / distance[current][city]) ** BETA
        for city in candidates
    ]

    return random.choices(candidates, weights=weights, k=1)[0]


def construct_route():
    route = [0]
    unvisited = set(range(1, n))

    while unvisited:
        next_city = select_next_city(route[-1], unvisited)
        route.append(next_city)
        unvisited.remove(next_city)

    return route


def route_distance(route):
    return sum(
        distance[route[i]][route[(i + 1) % n]]
        for i in range(n)
    )


def update_pheromone(routes):
    for i in range(n):
        for j in range(n):
            pheromone[i][j] *= (1 - RHO)

    for route, length in routes:
        deposit = Q / length

        for i in range(n):
            a = route[i]
            b = route[(i + 1) % n]
            pheromone[a][b] += deposit
            pheromone[b][a] += deposit


def main():
    global pheromone
    best_route = None
    best_distance = float("inf")

    for _ in range(ITERATIONS):
        routes = []

        for _ in range(NUM_ANTS):
            route = construct_route()
            length = route_distance(route)
            routes.append((route, length))

            if length < best_distance:
                best_route = route[:]
                best_distance = length

        update_pheromone(routes)

    result = [names[i] for i in best_route]
    result.append(result[0])

    print("TOURIST ROUTE OPTIMIZATION USING ACO")
    print("Best Route:", " -> ".join(result))
    print("Total Distance:", round(best_distance, 2), "units")


if __name__ == "__main__":
    main()
```
